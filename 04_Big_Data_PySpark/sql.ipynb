**Introduction to Spark SQL**

What is Spark SQL?

Spark SQL is a Spark module used to process structured data using SQL queries.

````python
from pyspark.sql import SparkSession
````

````python
s=SparkSession.builder.appName("test").getOrCreate()
emp=[(1,"sachin","IT",10000),
     (2,"saurav","HR",20000),
     (3,"virat","Finance",30000),
     (4,"dhoni","Admin",40000),
     (5,"rohit","IT",50000),
     (6,"shikhar","IT",60000),
     (7,"yuvraj","HR",70000),
     (8,"hardik","R&D",80000),
     (9,"kl rahul","R&D",90000),
     (10,"pant","Admin",100000)]

col=["id","name","Dept","salary"]
df=s.createDataFrame(emp,col)
df.show()
````

````python
df.createOrReplaceTempView("emp")
````

````python
r=s.sql("select * from emp")
r.show()
````

````python
s.sql("select * from emp where salary>50000").show()
````

````python
s.sql("select dept,avg(salary) from emp group by dept").show()
````

````python
s.sql("select count(*) as tot_emp from emp").show()
````

## Spark SQL

### 1. SparkSession

1.  **Definition**: `SparkSession` is the entry point to programming Spark with the Dataset and DataFrame API.
2.  **What it does**: It initializes a Spark application, providing a unified entry point to various Spark functionalities.
3.  **Why we use it**: It is required to create DataFrames, register DataFrames as tables, and execute SQL queries.
4.  **Syntax**: `SparkSession.builder.appName("YourAppName").getOrCreate()`
5.  **Simple example**:


````python
from pyspark.sql import SparkSession
s = SparkSession.builder.appName("my_spark_app").getOrCreate()
print("SparkSession created successfully!")
````

6.  **Expected output**:
```
SparkSession created successfully!
```


### 2. `createDataFrame()`

1.  **Definition**: A method in `SparkSession` to create a Spark DataFrame.
2.  **What it does**: It converts a collection of data (like a list of tuples or RDDs) into a structured DataFrame.
3.  **Why we use it**: To load data from various sources into Spark's optimized DataFrame structure for distributed processing and analysis.
4.  **Syntax**: `spark_session.createDataFrame(data, schema)`
5.  **Simple example**:


````python
sample_data = [("Alice", 1, "NY"), ("Bob", 2, "CA")]
schema = ["name", "id", "city"]
sample_df = s.createDataFrame(sample_data, schema)
sample_df.show()
````

6.  **Expected output**:
```
+-----+---+----+
| name| id|city|
+-----+---+----+
|Alice|  1|  NY|
|  Bob|  2|  CA|
+-----+---+----+
```


### 3. `createOrReplaceTempView()`

1.  **Definition**: A DataFrame method that creates or replaces a temporary view in the SparkSession's catalog.
2.  **What it does**: It registers a DataFrame as a temporary SQL table, making it queryable using Spark SQL.
3.  **Why we use it**: To allow users to interact with DataFrames using standard SQL syntax, simplifying data manipulation for SQL-savvy users.
4.  **Syntax**: `dataframe.createOrReplaceTempView("view_name")`
5.  **Simple example**:


````python
emp_data = [("sachin", 10000), ("rohit", 50000)]
emp_cols = ["name", "salary"]
emp_df = s.createDataFrame(emp_data, emp_cols)
emp_df.createOrReplaceTempView("sample_emp_view")
print("Temporary view 'sample_emp_view' created.")
````

6.  **Expected output**:
```
Temporary view 'sample_emp_view' created.
```


### 4. Spark SQL

1.  **Definition**: A Spark module for structured data processing.
2.  **What it does**: It allows you to query DataFrames and DataSets using SQL syntax, integrating relational processing with Spark's functional programming API.
3.  **Why we use it**: To leverage familiar SQL for data manipulation and analysis within the Spark ecosystem, making it accessible to a wider audience.
4.  **Syntax**: `spark_session.sql("SQL_QUERY")`
5.  **Simple example**:


````python
result_df = s.sql("SELECT name, salary FROM sample_emp_view")
result_df.show()
````

6.  **Expected output**:
```
+------+------+
|  name|salary|
+------+------+
|sachin| 10000|
| rohit| 50000|
+------+------+
```


### 5. `SELECT`

1.  **Definition**: The `SELECT` statement is used to retrieve data from one or more tables or views.
2.  **What it does**: It specifies which columns you want to fetch from the data source.
3.  **Why we use it**: To choose specific data columns for analysis, display, or further processing, rather than retrieving all columns.
4.  **Syntax**: `SELECT column1, column2, ... FROM table_name`
5.  **Simple example**:


````python
s.sql("SELECT name, Dept FROM emp").show()
````

6.  **Expected output**:
```
+--------+-------+

|  hardik|    R&D|
|kl rahul|    R&D|
|    pant|  Admin|
+--------+-------+
```


### 6. `WHERE`

1.  **Definition**: The `WHERE` clause is used to filter records based on a specified condition.
2.  **What it does**: It selects only those rows that satisfy the given criteria, returning a subset of the data.
3.  **Why we use it**: To narrow down the results and work with only the relevant data, making queries more efficient and results more focused.
4.  **Syntax**: `SELECT column1, ... FROM table_name WHERE condition`
5.  **Simple example**:


````python
s.sql("SELECT * FROM emp WHERE Dept = 'IT'").show()
````

6.  **Expected output**:
```
+---+-------+----+------+
| id|   name|Dept|salary|
+---+-------+----+------+
|  1| sachin|  IT| 10000|
|  5|  rohit|  IT| 50000|
|  6|shikhar|  IT| 60000|
+---+-------+----+------+
```


### 7. `GROUP BY`

1.  **Definition**: The `GROUP BY` clause groups rows that have the same values in specified columns into a summary row.
2.  **What it does**: It partitions the result set into groups based on the values in one or more columns, allowing aggregate functions to be applied to each group.
3.  **Why we use it**: To perform aggregate calculations (like sum, average, count, min, max) on groups of data, providing summarized insights rather than individual row details.
4.  **Syntax**: `SELECT column1, aggregate_function(column2) FROM table_name GROUP BY column1`
5.  **Simple example**:


````python
s.sql("SELECT Dept, COUNT(*) AS EmployeeCount FROM emp GROUP BY Dept").show()
````

6.  **Expected output**:
```
+-------+-------------+
|   Dept|EmployeeCount|
+-------+-------------+
|     HR|            2|
|Finance|            1|
|  Admin|            2|
|     IT|            3|
|    R&D|            2|
+-------+-------------+
```


### 8. `AVG()`

1.  **Definition**: `AVG()` is an SQL aggregate function that calculates the average value of a numeric column.
2.  **What it does**: It returns the arithmetic mean of all non-NULL values within a specified group or the entire dataset.
3.  **Why we use it**: To quickly find the average of a metric (e.g., average salary, average sales) for specific groups or for the overall data, providing a central tendency measure.
4.  **Syntax**: `AVG(column_name)`
5.  **Simple example**:


````python
s.sql("SELECT Dept, AVG(salary) AS AverageSalary FROM emp GROUP BY Dept").show()
````

6.  **Expected output**:
```
+-------+-------------+
|   Dept|AverageSalary|
+-------+-------------+
|     HR|      45000.0|
|Finance|      30000.0|
|  Admin|      70000.0|
|     IT|      40000.0|
|    R&D|      85000.0|
+-------+-------------+
```


### 9. `COUNT()`

1.  **Definition**: `COUNT()` is an SQL aggregate function that counts the number of rows or non-NULL values in a column.
2.  **What it does**: It returns the number of items in a specified group or the total number of rows in the result set.
3.  **Why we use it**: To get a total count of records, to count items within specific categories, or to determine the number of distinct values.
4.  **Syntax**: `COUNT(*)`, `COUNT(column_name)`, `COUNT(DISTINCT column_name)`
5.  **Simple example**:


````python
s.sql("SELECT COUNT(*) AS TotalEmployees FROM emp").show()
````

6.  **Expected output**:
```
+--------------+
|TotalEmployees|
+--------------+
|            10|
+--------------+
```


### 10. `AS` (Alias)

1.  **Definition**: The `AS` keyword is used to give a temporary name (an alias) to a column or a table.
2.  **What it does**: It renames a column or a table in the query's output, without changing the original schema.
3.  **Why we use it**: For clarity and brevity, especially when dealing with complex queries, aggregate functions, or long column/table names, to make the results more readable.
4.  **Syntax**: `SELECT column_name AS alias_name FROM table_name`
5.  **Simple example**:


````python
s.sql("SELECT name AS EmployeeName, Dept AS Department FROM emp").show()
````

6.  **Expected output**:
