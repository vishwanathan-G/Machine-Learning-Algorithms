### 1. `SparkSession`

**1. Definition:** `SparkSession` is the entry point to programming Spark with the DataFrame API.

**2. What it does:** It creates a session that allows you to interact with Spark, manage resources, and perform Spark operations.

**3. Why we use it:** It's the first step to initializing Spark in your application, providing a way to create DataFrames and access all Spark functionalities.

**4. Syntax:** `SparkSession.builder.appName("YourAppName").getOrCreate()`

**5. Simple example:**

````python
from pyspark.sql import SparkSession
s = SparkSession.builder.appName("test_revision").getOrCreate()
````

**6. Expected output:** This command initializes the Spark session. No direct output is displayed, but Spark is ready to be used.

### 2. `createDataFrame()`

**1. Definition:** A method of `SparkSession` used to create a DataFrame.

**2. What it does:** It converts a local Python collection (like a list of tuples or dictionaries) into a distributed Spark DataFrame.

**3. Why we use it:** To load small to medium-sized datasets from Python objects into Spark's optimized DataFrame structure for further processing.

**4. Syntax:** `spark_session.createDataFrame(data, schema_or_column_names)`

**5. Simple example:**

````python
emp_data = [(1, "sachin", 10000),
            (2, "saurav", 20000)]
emp_cols = ["id", "name", "salary"]

df_revision = s.createDataFrame(emp_data, emp_cols)
df_revision.show()
````

**6. Expected output:**
```
+---+------+------+
| id|  name|salary|
+---+------+------+
|  1|sachin| 10000|
|  2|saurav| 20000|
+---+------+------+
```

### 3. `show()`

**1. Definition:** A DataFrame action.

**2. What it does:** Displays the top `n` rows of the DataFrame in a tabular format.

**3. Why we use it:** Essential for inspecting the content of a DataFrame and verifying data after transformations.

**4. Syntax:** `dataframe.show(n=20, truncate=True, vertical=False)`

**5. Simple example:**

````python
df_revision.show(1)
````

**6. Expected output:**
```
+---+------+------+
| id|  name|salary|
+---+------+------+
|  1|sachin| 10000|
+---+------+------+
```

### 4. `select()`

**1. Definition:** A DataFrame transformation.

**2. What it does:** Selects a specified list of columns from the DataFrame.

**3. Why we use it:** To choose a subset of columns to work with, effectively reducing the DataFrame's width and focusing on relevant data.

**4. Syntax:** `dataframe.select("column1", "column2", ...)`

**5. Simple example:**

````python
df_revision.select("name", "salary").show()
````

**6. Expected output:**
```
+------+------+
|  name|salary|
+------+------+
|sachin| 10000|
|saurav| 20000|
+------+------+
```

### 5. `filter()`

**1. Definition:** A DataFrame transformation.

**2. What it does:** Filters rows based on a given condition, returning a new DataFrame containing only the rows that satisfy the condition.

**3. Why we use it:** To retrieve specific subsets of data from the DataFrame based on criteria, similar to a `WHERE` clause in SQL.

**4. Syntax:** `dataframe.filter(condition)`

**5. Simple example:**

````python
df_revision.filter(df_revision.salary > 10000).show()
````

**6. Expected output:**
```
+---+------+------+
| id|  name|salary|
+---+------+------+
|  2|saurav| 20000|
+---+------+------+
```

### 6. `col()`

**1. Definition:** A PySpark SQL function (`pyspark.sql.functions.col`).

**2. What it does:** It returns a `Column` object representing a column in a DataFrame.

**3. Why we use it:** It's used to reference columns in DataFrame operations when performing calculations, comparisons, or other expressions, especially within functions like `withColumn` or `filter` conditions.

**4. Syntax:** `from pyspark.sql.functions import col` then `col("column_name")`

**5. Simple example:**

````python
from pyspark.sql.functions import col

# Using col() in a filter operation
df_revision.filter(col("salary") == 10000).show()
````

**6. Expected output:**
```
+---+------+------+
| id|  name|salary|
+---+------+------+
|  1|sachin| 10000|
+---+------+------+
```

### 7. `withColumn()`

**1. Definition:** A DataFrame transformation.

**2. What it does:** Adds a new column or replaces an existing one with a specified value or expression.

**3. Why we use it:** To derive new features, perform calculations, or modify existing columns in a DataFrame, which is fundamental for data preparation.

**4. Syntax:** `dataframe.withColumn("new_column_name", expression_for_new_column)`

**5. Simple example:**

````python
df_with_bonus = df_revision.withColumn("Bonus_Revision", col("salary") + 5000)
df_with_bonus.show()
````

**6. Expected output:**
```
+---+------+------+
| id|  name|salary|Bonus_Revision|
+---+------+------+
|  1|sachin| 10000| 15000|
|  2|saurav| 20000| 25000|
+---+------+------+
```

### 8. `drop()`

**1. Definition:** A DataFrame transformation.

**2. What it does:** Removes one or more columns from a DataFrame.

**3. Why we use it:** To discard columns that are no longer needed, are redundant, or contain sensitive information, helping to streamline the DataFrame.

**4. Syntax:** `dataframe.drop("column_name1", "column_name2", ...)`

**5. Simple example:**

````python
df_dropped = df_with_bonus.drop("salary")
df_dropped.show()
````

**6. Expected output:**
```
+---+------+------------+
| id|  name|Bonus_Revision|
+---+------+------------+
|  1|sachin|       15000|
|  2|saurav|       25000|
+---+------+------------+
```
