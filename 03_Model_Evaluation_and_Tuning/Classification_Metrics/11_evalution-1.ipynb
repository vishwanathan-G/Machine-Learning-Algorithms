💡 SIMPLE EXAMPLE
y_true = [1,1,0,0,1,0,1,0]
y_pred = [1,0,0,0,1,1,1,0]

```python
y_true = [1,1,0,0,1,0,1,0] # True labels (Answer key)
y_pred = [1,0,0,0,1,1,1,0] # Predicted labels (Answer sheet)

       #TP, FN , TN , TN, TP, FP , TP, TN
```

TP (True Positive) → Correctly predicted YES
TN (True Negative) → Correctly predicted NO
FP (False Positive) → Wrongly predicted YES
FN (False Negative) → Missed YES


```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_true, y_pred)
print(cm)
```

[[3 1]
 [1 3]]

|            | Pred YES | Pred NO |
| ---------- | -------- | ------- |
| Actual YES | 3 (TP)   | 1 (FN)  |
| Actual NO  | 1 (FP)   | 3 (TN)  |


PRECISION

Formula:

Precision=
TP / FP + TP	​



```python
from sklearn.metrics import precision_score

print("Precision:", precision_score(y_true, y_pred))
```


👉 Out of all predicted YES, how many are correct?

TP = 3
FP = 1

👉 Precision = 3 / (3 + 1) = 0.75

RECALL

Formula:

Recall=
TP / TP + FN	​


```python
from sklearn.metrics import recall_score

print("Recall:", recall_score(y_true, y_pred))
```

👉 Out of actual YES, how many we correctly found?


TP = 3
FN = 1

👉 Recall = 3 / (3 + 1) = 0.75

F1 SCORE (BALANCE)

Formula:

F1=2 * 
Precision * Recall / Precision+Recall​


```python
from sklearn.metrics import f1_score

print("F1 Score:", f1_score(y_true, y_pred))
```
