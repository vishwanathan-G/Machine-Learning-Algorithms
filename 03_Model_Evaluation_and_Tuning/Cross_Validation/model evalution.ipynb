 Holdout Method (Regression)
Splits into 80% train / 20% test.



```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import pandas as pd

# Load data (petal length ➝ petal width)
df=pd.read_csv(r"iris.csv")


X=df[["sepal_length","sepal_width","petal_length","petal_width"]]
y=df["species"]


# Holdout split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8) #model selection

# Model
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, y_train)

# Evaluation
y_pred = model.predict(X_test)
print("Accuracy Score:", accuracy_score(y_test, y_pred))

```

MSE: 0.047.

Good for a quick estimate, but performance may vary across runs.

Bootstrap Method (Regression)
Trains 100 models on resampled datasets.

```python
from sklearn.datasets import load_iris
from sklearn.linear_model import LinearRegression
from sklearn.utils import resample
from sklearn.metrics import mean_squared_error
import numpy as np

# Load data
iris = load_iris()
X = iris.data[:, 2].reshape(-1, 1)
y = iris.data[:, 3]

# Bootstrap
n_iterations = 10000
mse_list = []

for _ in range(n_iterations): # Loop for bootstrap resampling - 10000 iterations
    X_resample, y_resample = resample(X, y)
    model = LinearRegression()
    model.fit(X_resample, y_resample)
    y_pred = model.predict(X)
    mse = mean_squared_error(y, y_pred) # Calculate MSE for each bootstrap sample - 100000 values
    mse_list.append(mse) #average MSE for 10000 iterations

print(mse_list[:10])
print("Bootstrap Avg. MSE:", np.mean(mse_list))
```

Avg. MSE: ~0.042–0.045

Slightly worse than Holdout since it averages over many noisy models, but gives a robust error estimate.

K-Fold CV (Regression)
Trains on 5 different train-test splits.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import KFold, cross_val_score
from sklearn.linear_model import LinearRegression
import numpy as np

# Load data
iris = load_iris()
X = iris.data[:, 2].reshape(-1, 1)
y = iris.data[:, 3]

# K-Fold
model = LinearRegression()
kf = KFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_val_score(model, X, y, cv=kf)

print(scores)
print("K-Fold Cross-Validation MSE:", scores.mean())
```

```python
scores
```

MSE: ~0.045

Most stable and unbiased estimate among all regression methods.

🔹Lazy Learner – KNN


```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

# Load data (for classification)
iris = load_iris()
X = iris.data[:, 2].reshape(-1, 1)
y = iris.target

# Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Lazy learner
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)

# Accuracy
print("Lazy Learner Accuracy (KNN):", knn.score(X_test, y_test))
```

Accuracy: ~97.7%

Sensitive to feature scale & choice of k.

Performance lower than Decision Tree.

------------------

🔹 Eager Learner – Decision Tree


```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

# Load data (same as KNN)
iris = load_iris()
X = iris.data[:, 2].reshape(-1, 1)
y = iris.target

# Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Eager learner
tree = DecisionTreeClassifier(random_state=42)
tree.fit(X_train, y_train)

# Accuracy
print("Eager Learner Accuracy (Decision Tree):", tree.score(X_test, y_test))
```

Accuracy: ~95.5%

Better classification performance using a single feature.

Might overfit with more complex trees or less data.

RECOMMENDATION

| Use Case                       | Best Method       |
| ------------------------------ | ----------------- |
| Small data + want stable error | **K-Fold CV**     |
| Real-time prediction           | **Holdout**       |
| Model uncertainty analysis     | **Bootstrap**     |
| Simple classification task     | **Decision Tree** |
| Distance-based classification  | **KNN** (Lazy)    |


```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
import numpy as np
import pandas as pd

# Load the Wine dataset
wine = load_wine()

# Convert to a DataFrame for better visualization
wine_df = pd.DataFrame(wine.data, columns=wine.feature_names)

# Show the first 5 rows of the dataset
#print(wine_df.head())

knn_scores = []
rf_scores = []

for seed in range(100):
    wine = load_wine()
    X = wine.data
    y = wine.target

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=seed)

    knn = KNeighborsClassifier(n_neighbors=3)
    rf = RandomForestClassifier(n_estimators=100, max_depth=10, random_state=seed)

    knn.fit(X_train, y_train)
    rf.fit(X_train, y_train)

    knn_scores.append(knn.score(X_test, y_test))
    rf_scores.append(rf.score(X_test, y_test))

print("Avg KNN Accuracy on Wine Dataset:", np.mean(knn_scores))
print("Avg Random Forest Accuracy on Wine Dataset:", np.mean(rf_scores))
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.utils import shuffle
import numpy as np

# Load data (petal length ➝ petal width)
iris = load_iris()
X = iris.data[:, 2].reshape(-1, 1)  # Petal length
y = iris.data[:, 3]                 # Petal width

mse_list = []

# Run 100 different train-test splits
for _ in range(100):
    # Shuffle data for different splits
    X_shuf, y_shuf = shuffle(X, y, random_state=None)
    X_train, X_test, y_train, y_test = train_test_split(X_shuf, y_shuf, test_size=0.2)

    # Model training
    model = LinearRegression()
    model.fit(X_train, y_train)

    # Prediction and MSE calculation
    y_pred = model.predict(X_test)
    mse = mean_squared_error(y_test, y_pred)
    mse_list.append(mse)

# Final result
average_mse = np.mean(mse_list)
print("Average MSE over 100 runs:", average_mse)
```
