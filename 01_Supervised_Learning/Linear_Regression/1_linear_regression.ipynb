```python
df = pd.read_csv('houseprice.csv')
df
```

```python
%matplotlib inline
plt.xlabel('area')
plt.ylabel('price')
plt.plot(df.area,df.price,color='red',marker='p')
```

```python
x = df[['area']]
x
```

```python
y = df[['price']]
y
```

```python
from sklearn import linear_model
```

```python
# Create linear regression object
model = linear_model.LinearRegression()
```

```python
model.fit(x,y) #fit - Training
```

**(1) Predict price of a home with area = 3300 sqr ft**

```python
model.predict([[5000]])
```

```python
model.coef_
```

```python
model.intercept_
```
