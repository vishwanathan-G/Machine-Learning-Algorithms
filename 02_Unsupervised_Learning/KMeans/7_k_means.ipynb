```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

```

```python
# Step 1: Original Data
data = {
    'X': [2, 2, 8, 5, 7, 6, 1, 4],
    'Y': [10, 5, 4, 8, 5, 4, 2, 9]
}
df = pd.DataFrame(data)
df
```

```python
# Step 2: Apply KMeans Clustering (no optional params)
kmeans = KMeans(n_clusters=4)
df['Cluster'] = kmeans.fit_predict(df[['X', 'Y']])
centroids = kmeans.cluster_centers_
centroids
```

```python
# Step 3: Accept New Point Input from User
x = float(input("Enter X coordinate of the new point: "))
y = float(input("Enter Y coordinate of the new point: "))
new_point = [[x, y]]

# Step 4: Predict the Cluster for New Point
cluster_label = kmeans.predict(new_point)[0]
print(f"The point ({x}, {y}) belongs to Cluster {cluster_label}")
```

```python
# Step 5: Plot Everything
plt.figure(figsize=(5, 5))
colors = ['red', 'green', 'blue']
#,'hotpink','indigo'

# Plot each cluster
for i in range(3):
    cluster_points = df[df['Cluster'] == i]
    plt.scatter(cluster_points['X'], cluster_points['Y'], 
                label=f'Cluster {i}', color=colors[i])

# Plot centroids
plt.scatter(centroids[:, 0], centroids[:, 1], 
            s=200, marker='X', c='black', label='Centroids')

# Plot new input point
plt.scatter(x, y, c='yellow', s=150, edgecolor='black', 
            marker='*', label='New Point')

# Add cluster label text near the new point
plt.text(x+0.2 , y+0.2, f'Cluster {cluster_label}', fontsize=12, color='black', weight='bold')

# Final plot settings
plt.title("K-Means Clustering with New Input Point")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)
plt.show()
```
